```python
import streamlit as st

st.title("EKSU CGPA / GPA CALCULATOR")

# Student info
name = st.text_input("Input your full name")
department = st.text_input("Input your department")
matric = st.text_input("Input your Matric No")
faculty = st.text_input("Input your Faculty")

total_point = 0
total_unit = 0

semester = st.number_input("How many semesters are you calculating?", min_value=1, step=1)

if semester:
    for calculation in range(int(semester)):

        st.subheader(f"Semester {calculation + 1}")

        courses = st.number_input(
            f"Number of courses for semester {calculation + 1}",
            min_value=1,
            step=1,
            key=f"course_count_{calculation}"
        )

        for session in range(int(courses)):

            course = st.text_input(
                f"Course {session + 1}",
                key=f"course_{calculation}_{session}"
            )

            grade = st.text_input(
                f"Grade for {course}",
                key=f"grade_{calculation}_{session}"
            ).upper()

            unit = st.number_input(
                f"Unit for {course}",
                min_value=1,
                step=1,
                key=f"unit_{calculation}_{session}"
            )

            if grade == "A":
                point = 5
            elif grade == "B":
                point = 4
            elif grade == "C":
                point = 3
            elif grade == "D":
                point = 2
            elif grade == "E":
                point = 1
            else:
                point = 0

            total_point += point * unit
            total_unit += unit

    if total_unit > 0:
        cgpa = total_point / total_unit

        st.subheader("RESULT")

        if cgpa > 5.00:
            st.error("Invalid Input.")
        else:
            st.write("Name:", name.title())
            st.write("Matric No:", matric)
            st.write("Department:", department.title())
            st.write("Faculty:", faculty.title())
            st.write("CGPA / GPA:", round(cgpa, 2))

            if cgpa >= 4.50:
                st.success("First Class Honours!")
            elif cgpa >= 3.50:
                st.success("Second Class Upper Honours!")
            elif cgpa >= 2.50:
                st.info("Second Class Lower Honours!")
            elif cgpa >= 1.50:
                st.warning("Third Class Honours")
            else:
                st.error("FAILED")
```
